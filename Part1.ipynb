```python
import csv
import math
import pandas as pd #only for loading data
```

```python
df = pd.read_csv('clean_final.csv')   #clean_final.csv used for moel training
```

```python
df
```

```python
df = df.drop(columns = ['Unnamed: 0'])
```

```python
X = df.drop('Price', axis =1)
y = df['Price']
y = [math.log(v) for v in y]
```

```python
X
```

```python
#Spliting data
split = int(0.8 * len(X))
X_train = X[0:split]
X_test = X[split:]
y_train = y[0:split]
y_test = y[split:]
```

```python
X_train
```

```python
#feature Scaling

#We need mean and standard deviation

#mean ka calculation
  #number of features
m = len(X_train)  #total number of rows in train data
n = 19 #number of features in train data
means =[]
for i in range (0,n):
    sum=0
    for j in range (0,m):
        sum += X_train.iloc[j,i]
    sum=sum/m
    means.append(sum)
```

```python
#calculating standard deviation
stds = []
for i in range(0, n):
    variance_sum = 0
    for j in range(0, m):
        variance_sum += (X_train.iloc[j, i] - means[i]) ** 2

    variance = variance_sum / m
    std = variance ** 0.5
    if std == 0:
        std = 1
    stds.append(std)
```

```python
# Applying scaling to train set
X_train_scaled = []
for j in range(0, m):
    row = []
    for i in range(0, n):
        row.append((X_train.iloc[j, i] - means[i]) / stds[i])
    X_train_scaled.append(row)
```

```python
m_test = len(X_test)
X_test_scaled = []
for j in range(0, m_test):
    row = []
    for i in range(0, n):
        row.append((X_test.iloc[j, i] - means[i]) / stds[i])
    X_test_scaled.append(row)
```

```python
y_train = list(y_train)
y_test  = list(y_test)
```

```python
#Gradient descent... help of AI was taken while coding, but I know the logic and understand the code
import time
weights = [0.0] * n   # initialising weights
bias    = 0.0         # bias
 
learning_rate  = 0.01
iterations     = 1000
 
cost_history      = []   # to store cost at every 100 iterations
iteration_history = []   # to store iteration number
 

start_time = time.time()
 
for iteration in range(iterations):
 
    # Computing predictions
    predictions = []
    for i in range(m):
        prediction = bias
        for j in range(n):
            prediction += weights[j] * X_train_scaled[i][j]
        predictions.append(prediction)
 
    # Computing cost (MSE)
    cost = 0
    for i in range(m):
        error = predictions[i] - y_train[i]
        cost += error * error
    cost = cost / (2 * m)
 
    # Computing gradients
    bias_gradient = 0
    for i in range(m):
        bias_gradient += (predictions[i] - y_train[i])
    bias_gradient = bias_gradient / m
 
    weight_gradients = [0.0] * n
    for j in range(n):   # calculating for each feature
        for i in range(m):   # for each training example
            weight_gradients[j] += (predictions[i] - y_train[i]) * X_train_scaled[i][j]
        weight_gradients[j] = weight_gradients[j] / m
 
    # Updating bias and weights
    bias = bias - learning_rate * bias_gradient
    for j in range(n):
        weights[j] = weights[j] - learning_rate * weight_gradients[j]
 
    # Storing cost every 100 iterations for plotting
    if iteration % 100 == 0:
        cost_history.append(cost)
        iteration_history.append(iteration)
        print(f"  Iteration {iteration:4d} | Cost: {cost:.6f}")
 
convergence_time = time.time() - start_time
print(f"\nConvergence time: {convergence_time:.4f} seconds")
 
```

```python
#Evaluation... took assistance of AI in code and also debugging
import math
from decimal import Decimal, getcontext
import numpy as np
getcontext().prec = 50
 
# Computing predictions on test set
test_predictions = []
for i in range(m_test):
    prediction = bias
    for j in range(n):
        prediction += weights[j] * X_test_scaled[i][j]
    test_predictions.append(prediction)
 
# r2 Score
m_test = int(len(y_test))
y_mean = np.sum(y_test) / m_test
 
tss = 0   # Total Sum of Squares
for i in range(m_test):
    tss += (y_test[i] - y_mean) ** 2
 
rss = 0   # Residual Sum of Squares
for i in range(m_test):
    rss += (y_test[i] - test_predictions[i]) ** 2
 
r2_score = 1 - (rss / tss)
print(f"R2 Score : {r2_score:.4f}")
 
# MAE and RMSE on log scale
mae = 0
for i in range(m_test):
    error = y_test[i] - test_predictions[i]
    if error < 0:
        error = -error
    mae += error
mae = mae / m_test
print(f"MAE (log scale)  : {mae:.4f}")

# RMSE on log scale
mse = 0
for i in range(m_test):
    error = y_test[i] - test_predictions[i]
    mse += error * error
mse = mse / m_test
rmse = mse ** 0.5
print(f"RMSE (log scale) : {rmse:.4f}")
```

```python
import matplotlib.pyplot as plt

#plotting cost curve
plt.figure(figsize=(10, 6))
plt.plot(iteration_history, cost_history, 'b-', linewidth=2, label='Cost')
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Cost Function During Gradient Descent for Part 1')
plt.grid(True, alpha=0.3)
plt.legend()
plt.savefig("Curve_Part1")
plt.show()
```

```python
mae_actual = math.exp(mae) - 1 #Coversion of mae and rmse
rmse_actual = math.exp(rmse) - 1

print(f"MAE (actual price) : ${mae_actual * 1000000:,.0f}")
print(f"RMSE (actual price) : ${rmse_actual * 1000000:,.0f}")
```
